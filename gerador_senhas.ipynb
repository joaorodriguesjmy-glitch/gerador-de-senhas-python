```python
import secrets
import string


def gerar_senha(tamanho=12, usar_maiusculas=True, usar_numeros=True, usar_simbolos=True):
    caracteres = string.ascii_lowercase

    if usar_maiusculas:
        caracteres += string.ascii_uppercase
    if usar_numeros:
        caracteres += string.digits
    if usar_simbolos:
        caracteres += "!@#$%^&*()?"

    senha = "".join(secrets.choice(caracteres) for _ in range(tamanho))
    return senha


print("=== GERADOR DE SENHAS SEGURAS ===")

try:
    tamanho = int(input("Digite o tamanho da senha (ex: 12): "))
    if tamanho < 4:
        print("Tamanho muito curto! Usando o mínimo recomendado de 8 caracteres.")
        tamanho = 8

    maiusculas = input("Incluir letras maiúsculas? (s/n): ").strip().lower() == "s"
    numeros = input("Incluir números? (s/n): ").strip().lower() == "s"
    simbolos = input("Incluir símbolos especiais? (s/n): ").strip().lower() == "s"

    senha_final = gerar_senha(tamanho, maiusculas, numeros, simbolos)
    print("\nSua senha gerada é:", senha_final)

except ValueError:
    print("Por favor, digite apenas números válidos para o tamanho.")
```
